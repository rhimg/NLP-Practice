```python
#Define corpus
corpus = """Artificial Intelligence (AI) is transforming the world in 2026. 
Machine Learning, Deep Learning, and Natural Language Processing (NLP) are key domains of AI.

In recent years, models like GPT-4, BERT, and LLaMA-2 have revolutionized text generation and understanding.
Companies such as OpenAI, Google, and Meta are investing billions of dollars in AI research.

Dr. A.P.J. Abdul Kalam once said, "Dream, dream, dream. Dreams transform into thoughts and thoughts result in action."

Today’s AI systems can analyze 1,000+ documents in seconds.
However, challenges remain — data privacy, bias, and model interpretability.

If you want to learn more, visit https://www.example-ai-research.org 
or contact research.team@example.com.

AI-powered systems are used in healthcare, finance, e-commerce, and autonomous vehicles.
Is AI the future? Absolutely! But ethical AI development is essential.
"""
```

```python
corpus
```

```python
print(corpus)
```

```python
## Tokenization
## Paragraphs --> Sentence
from nltk.tokenize import sent_tokenize
```

```python
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
documents = sent_tokenize(corpus)
```

```python
type(documents)
```

```python
for sentence in documents:
    print(sentence)
```

```python
## Tokenization
## Paragraph --> words
## Sentence --> words
from nltk.tokenize import word_tokenize
```

```python
word_tokenize(corpus)
```

```python
for sentence in documents:
    print(word_tokenize(sentence))
```

```python
from nltk.tokenize import wordpunct_tokenize
```

```python
wordpunct_tokenize(corpus)
```

```python
from nltk.tokenize import TreebankWordTokenizer
```

```python
tokenizer = TreebankWordTokenizer()
```

```python
tokenizer.tokenize(corpus)
```
